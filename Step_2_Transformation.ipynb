```python
import pandas as pd
```

```python
X_train = pd.read_csv("/Users/vijaysabhavat/Documents/Airline_ML_Project_001/Data/CleanedData/X_train.csv")

X_test = pd.read_csv("/Users/vijaysabhavat/Documents/Airline_ML_Project_001/Data/CleanedData/X_test.csv")

X_train
```

```python
# Transformation encoding
# Ordinal Col
# one hote col
# Numerical Col

numericalCol = ['Total_Stops','DurHrs_Mins','D_Day','D_Month','D_Year','D_Hour','D_Minute','A_Hour','A_Minute','A_Day','A_Month','A_Year']
OneHotCol =['Airline','Source','Destination'] 
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler

OneHotEncoding_01 = ("One_Hot_Encoder",OneHotEncoder(sparse_output=False, drop="first", handle_unknown="ignore"),OneHotCol,)
Scaling = ("Numerical_Coluns", StandardScaler(), numericalCol)

transform_Obj = ColumnTransformer(transformers=[OneHotEncoding_01,Scaling]).set_output(transform='pandas')

transform_Obj
```

```python
X_train_trans = transform_Obj.fit_transform(X_train)
X_train_trans
```

```python
X_train_trans.to_csv("/Users/vijaysabhavat/Documents/Airline_ML_Project_001/Data/Trans_Data/X_train_trans.csv")
```

```python
import joblib
joblib.dump(transform_Obj,"/Users/vijaysabhavat/Documents/Airline_ML_Project_001/Transform.pkl")
```

```python
TransFile = joblib.load("/Users/vijaysabhavat/Documents/Airline_ML_Project_001/Transform.pkl")
TransFile
```

```python
X_test_trans = TransFile.fit_transform(X_test)
X_test_trans.to_csv("/Users/vijaysabhavat/Documents/Airline_ML_Project_001/Data/Trans_Data/X_test_trans.csv")
```
